```python
import matplotlib.pyplot as plt
import numpy as np

x = np.array([2023,2024,2025,2026])
y1 = np.array([15,50,30,20])
y2 = np.array([17,90,28,50])
y3 = np.array([10,30,22,100])

line_style = dict (
    marker = "v",
    markeredgecolor="red",
    markerfacecolor="red",
    markersize=8
)


plt.plot(x,y1, color="green", **line_style)
plt.plot(x,y2, color="red", **line_style)
plt.plot(x,y3, color="blue", **line_style)

plt.xlabel("Year")
plt.ylabel("scores")
plt.title("Total Scores")
plt.xticks(x)
plt.tight_layout()



plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

catergories = np.array(["USA", "Ghana", "UK", "Spain", "Italy", "France"])
values = np.array([1,2,3,4,5,6])

plt.bar(catergories,values)
plt.show()
```

```python
categories = np.array(["Freshmen", "Sophomores", "Junior", "Seniors"])
students = np.array([300, 250, 275, 225])
colors = ["red", "yellow", "blue", "green"]

plt.pie(students, labels = categories,
                  shadow=True,
                  explode=[0,0,0.1,0],
                  autopct="%1.1f%%",
                  colors=colors,
                  startangle=90)

plt.show()
```

```python
x1 = [0,1,1,2,3,4,5,6,7,7,8]
y1= [55, 60,65,63,68,70,75,78,82,85,87]

x2 = [0,1,4,2,1,4,9,6,8,7,8]
y2= [50, 60,30,63,68,40,75,100,82,10,87]

plt.scatter(x1,y1, color="green",
                   alpha=0.5,
                   s = 200,
                   label="Class A")

plt.scatter(x2,y2, color="blue",
                   alpha=0.5,
                   s = 200,
                   label="Class B")


plt.xlabel("Hours Studied")
plt.ylabel("Grade")
plt.title("Test Scores")
plt.legend()

plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

scores = np.random.normal(loc=80, scale=10, size=100)
scores = np.clip(scores, 0,100)

plt.hist(scores, bins=10,
                 color="red",
                 edgecolor="black")

plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.array([1,2,3,4,5])

figure, axes = plt.subplots(2,2)

axes[0,0].plot(x, x*2, color="red")
axes[0,0].set_title("x*2")

axes[0,1].plot(x, x**2, color="blue")
axes[0,0].set_title("x**2")


axes[1,0].plot(x, x**3, color="green")
axes[1,0].set_title("x**3")


axes[1,1].plot(x, x**3, color="purple")
axes[1,1].set_title("x**4")

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.array([2023,2024,2024,2025])
y1 = np.array([15,50,30,20])
y2 = np.array([17,90,28,50])
y3 = np.array([10,30,22,100])

line_style = dict( marker="v", 
              markersize=10,
              markerfacecolor="red",
              markeredgecolor="red",
              linestyle="solid",
              linewidth=4,)

plt.plot(x,y1, color="red", **line_style)
plt.plot(x,y2, color="green", **line_style)
plt.plot(x,y3, color="blue", **line_style)

plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.array([2023,2024,2024,2025])
y1 = np.array([15,50,30,20])
y2 = np.array([17,90,28,50])
y3 = np.array([10,30,22,100])

plt.title("Class size", fontsize=25,
                        family="Arial",
                        fontweight="bold",
                        color="#2d4cfc")

plt.xlabel("Year", fontsize=25,
                   family="Arial",
                   fontweight="bold",
                   color="#2d4cfc"
                   )


plt.ylabel("Students", fontsize=25,
                   family="Arial",
                   fontweight="bold",
                   color="#2d4cfc"
                   )

plt.tick_params(axis="both",
                color="#2d4cfc")

plt.plot(x,y1)
plt.plot(x,y2)
plt.plot(x,y3)

plt.xticks(x)

plt.show()
```

```python
# grid lines 

import matplotlib.pyplot as plt
# import numpy as np

x = [1,2,3,4,5]
y = [6,7,8,9,10]

plt.grid(axis="y",
         linewidth=2,
         color="lightgray",
         linestyle="dashed")
plt.plot(x,y)

plt.show()
```

```python
# bar charts

catergories = np.array(["Grains", "Bread", "Vegetables", "Protein", "Dairy", "Sweets"])
values = np.array ([4,3,2,5,3,1])

plt.bar(catergories,values, color="skyblue")
# plt.barh(catergories,values, color="skyblue")

plt.title("Daily Consumption")
plt.xlabel("Food")
plt.ylabel("Quantity")
plt.show
```
